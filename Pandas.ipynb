```python
import pandas as pd
```

```python
data = {
    "Title": ["Kitab A", "Kitab B", "Kitab C", "Kitab D", "Kitab E", "Kitab F"],
    "Author": ["Ali", "Veli", "Murad", "Aysel", "Nigar", "Elvin"],
    "PublicationYear": [2010, 2018, 2020, 2012, 2019, 2023],
    "Price": [15, 25, 30, 12, 20, 35]
}

df = pd.DataFrame(data)

print(df)

print("Orta qiymət:", df["Price"].mean())

print(df[df["PublicationYear"] > 2015])

print(df.sort_values(by="Price"))
```

```python
import pandas as pd

# Create a sample orders.csv file since it does not exist
sample_data = {
    "Customer": ["Ali", "Veli", "Ali", "Aysel", "Nigar", "Ali", "Elvin", "Veli"],
    "Amount": [120, 150, 80, 200, 50, 300, 110, 95]
}
pd.DataFrame(sample_data).to_csv("orders.csv", index=False)

# Now read the csv and perform the operations
df = pd.read_csv("orders.csv")

print(df.head(10))

print(df.groupby("Customer").size())

print("Maksimum məbləğ:", df["Amount"].max())

print("Minimum məbləğ:", df["Amount"].min())

print("Ümumi məbləğ:", df["Amount"].sum())
```

```python
data = {
    "Product": ["Alma", "Banan", "Toyuq", "Düyü", "Çörək",
                "Balıq", "Kartof", "Pendir", "Yumurta", "Ət"],
    "Category": ["Meyvə", "Meyvə", "Ət", "Taxıl", "Taxıl",
                 "Ət", "Tərəvəz", "Süd məhsulu", "Süd məhsulu", "Ət"],
    "Calories": [80, 120, 350, 400, 280, 320, 150, 450, 200, 500],
    "Protein": [1, 2, 30, 8, 7, 25, 3, 20, 13, 35]
}

df = pd.DataFrame(data)

print(df)

print(df[df["Calories"] > 300])

print(df.groupby("Category")["Protein"].mean())

print(df.sort_values(by="Calories", ascending=False))
```

```python
data = {
    "Name": ["Ali", "Veli", "Murad", "Ali", "Aysel",
             "Veli", "Murad", "Aysel"],
    "Project": ["A", "B", "A", "C", "B", "A", "C", "B"],
    "Hours": [10, 15, 20, 12, 18, 8, 14, 10]
}

df = pd.DataFrame(data)

print(df)

print(df.groupby("Name")["Hours"].sum())

print(df.groupby("Project")["Hours"].sum())

employee_hours = df.groupby("Name")["Hours"].sum()

print("Ən çox işləyən əməkdaş:", employee_hours.idxmax())
print("İş saatı:", employee_hours.max())
```

```python
data = {
    "Movie": ["Avatar", "Titanic", "Avatar", "Joker",
              "Titanic", "Joker", "Avatar", "Batman",
              "Batman", "Titanic", "Joker", "Batman"],
    "City": ["Baku", "Ganja", "Baku", "Sumqayit",
             "Baku", "Ganja", "Ganja", "Baku",
             "Sumqayit", "Ganja", "Baku", "Sumqayit"],
    "TicketsSold": [100, 80, 120, 90, 110, 70,
                    60, 130, 100, 50, 140, 90]
}

df = pd.DataFrame(data)

print(df)

print(df.groupby("Movie")["TicketsSold"].sum())

print(df.groupby("City")["TicketsSold"].sum())

movie_sales = df.groupby("Movie")["TicketsSold"].sum()

print("Ən çox bilet satılan film:", movie_sales.idxmax())
print("Satılan biletlərin sayı:", movie_sales.max())
```
