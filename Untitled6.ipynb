```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    days,
    temperature,
    marker="o",
    linestyle="--",
    linewidth=2
)

plt.title("7 Günlük Temperatur Qrafiki")
plt.xlabel("Günlər")
plt.ylabel("Temperatur (°C)")

plt.grid(True)

for i in range(len(days)):
    plt.text(days[i], temperature[i], str(temperature[i]),
             ha="center", va="bottom")

plt.show()
```

```python
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]
```

```python
colors = ["green", "orange", "blue", "purple", "red"]
```

```python
plt.figure(figsize=(10, 5))

bars = plt.barh(languages, students, color=colors)

bars[0].set_color("gold")

plt.title("Proqramlaşdırma Dilləri üzrə Tələbə Sayı")
plt.xlabel("Tələbə sayı")
plt.ylabel("Proqramlaşdırma dilləri")

for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height()/2,
             str(int(width)), va="center")

plt.show()

plt.figure(figsize=(10, 5))

bars = plt.bar(languages, students, color=colors)

bars[0].set_color("gold")

plt.title("Proqramlaşdırma Dilləri üzrə Tələbə Sayı")
plt.xlabel("Proqramlaşdırma dilləri")
plt.ylabel("Tələbə sayı")

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.5,
             str(int(height)), ha="center")

plt.show()
```

```python
import matplotlib.pyplot as plt
```

```python
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]
```

```python
fig = plt.figure(figsize=(12, 6))

ax1 = fig.add_axes([0.08, 0.15, 0.55, 0.75])

scatter = ax1.scatter(
    study_hours,
    scores,
    c=ages,
    cmap="viridis",
    marker="o",
    s=100,
    edgecolors="black"
)

max_index = scores.index(max(scores))

ax1.scatter(
    study_hours[max_index],
    scores[max_index],
    color="red",
    marker="*",
    s=250,
    label="Ən yüksək nəticə"
)

ax1.annotate(
    "95 bal",
    (study_hours[max_index], scores[max_index]),
    xytext=(10, 10),
    textcoords="offset points"
)

ax1.set_title("Dərs Saatları və İmtahan Nəticələri")
ax1.set_xlabel("Study Hours")
ax1.set_ylabel("Exam Score")

ax1.grid(True)
ax1.legend()

cbar = fig.colorbar(scatter, ax=ax1)
cbar.set_label("Age")

ax2 = fig.add_axes([0.72, 0.25, 0.23, 0.55])

ax2.bar(
    range(1, len(scores) + 1),
    scores,
    color="skyblue"
)

ax2.set_title("Exam Scores")
ax2.set_xlabel("Student")
ax2.set_ylabel("Score")

plt.show()
```
